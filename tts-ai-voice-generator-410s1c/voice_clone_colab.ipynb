# Free AI Voice Generator (Voice Cloning)

Runs the open-source XTTS-v2 model for free on Google Colab — no credits, no API keys.

**How to use:**
1. In the menu: Runtime → Change runtime type → select **T4 GPU** (free).
2. Run each cell below in order (▶ button).
3. Edit the `TEXT` in the last cell to whatever you want spoken, run it, and the MP3 downloads automatically.

Your voice sample is loaded straight from this repo — nothing to upload.

Cells 1 and 2 only need to run once per session. After that, just edit `TEXT` in cell 3 and re-run it — every run is a new free MP3.

```python
# 1. Install (takes ~2 minutes)
!pip install -q coqui-tts pydub
!apt-get -qq install -y ffmpeg
```

```python
# 2. Get the voice sample from the repo and load the model (~2 GB download, one time)
import os
os.environ["COQUI_TOS_AGREED"] = "1"

!wget -q -O voice_sample.mp3 https://raw.githubusercontent.com/moraj98/app-for-voice-1/claude/tts-ai-voice-generator-410s1c/voice_samples/mohit_sample.mp3

from TTS.api import TTS
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to("cuda")
print("Model ready!")
```

```python
# 3. Generate! Edit TEXT and LANGUAGE, then run this cell. Re-run as many times as you like — always free.
TEXT = """क्रिकेट में 'हिट विकेट' तब किसे माना जाता है जब बल्लेबाज खुद अपने ही बल्ले या शरीर से गिल्लियाँ गिरा दे?"""
LANGUAGE = "hi"  # "hi" for Hindi, "en" for English

from pydub import AudioSegment
from google.colab import files

tts.tts_to_file(text=TEXT, speaker_wav="voice_sample.mp3", language=LANGUAGE, file_path="out.wav")
AudioSegment.from_wav("out.wav").export("ai_voice.mp3", format="mp3", bitrate="192k")
files.download("ai_voice.mp3")
```
